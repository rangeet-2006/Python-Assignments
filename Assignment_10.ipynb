```python
# 1. WAP in Python to create a set and add elements using a user-defined function. Also, access the elements.
def add_element(s, item):
    s.add(item)

s = set()
n = int(input("Enter number of elements: "))

for i in range(n):
    x = input("Enter element: ")
    add_element(s, x)
print("Set:", s)

print("Elements of set:")
for i in s:
    print(i)
```

```python
# 2. WAP in Python to demonstrate the use of a set to remove duplicate elements from a list.
n = int(input("Enter number of elements: "))
L = []

for i in range(n):
    L.append(input("Enter element: "))
print("Original List:", L)

S = set(L)
print("List after removing duplicates:", list(S))
```

```python
# 3. WAP in Python to create a frozenset from a given list of elements.
n = int(input("Enter number of elements: "))
L = []

for i in range(n):
    L.append(input("Enter element: "))

fs = frozenset(L)
print("List:", L)
print("Frozen Set:", fs)
```

```python
# 4. WAP in Python to create a set of prime numbers up to a specified number.
n = int(input("Enter a number: "))
s = set()

for i in range(2, n+1):
    flag = 1
    for j in range(2, i):
        if i % j == 0:
            flag = 0
            break
    if flag == 1:
        s.add(i)
print("Prime Numbers Set:", s)
```

```python
# 5. WAP in Python to find the union and intersection of two sets.
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}

print("Union:", s1.union(s2))
print("Intersection:", s1.intersection(s2))
```

```python
# 6. WAP in Python to check if a set is a subset or superset of another set using user-defined functions.
# Check subset and superset using user-defined functions

def check_subset(a, b):
    if a.issubset(b):
        print("s1 is a subset of s2")
    else:
        print("s1 is not a subset of s2")

def check_superset(a, b):
    if a.issuperset(b):
        print("s1 is a superset of s2")
    else:
        print("s1 is not a superset of s2")

s1 = {1, 2}
s2 = {1, 2, 3, 4}

check_subset(s1, s2)
check_superset(s1, s2)
```

```python
# 7. WAP in Python to join multiple sets into one set using the update() method.
s1 = set(map(int, input("Enter elements of set 1: ").split()))
s2 = set(map(int, input("Enter elements of set 2: ").split()))
s3 = set(map(int, input("Enter elements of set 3: ").split()))

s1.update(s2, s3)
print("Joined Set:", s1)
```

```python
# 8. WAP in Python to slice elements from a frozenset by first converting it to a list
n = int(input("Enter number of elements: "))
L = []

for i in range(n):
    L.append(int(input("Enter element: ")))

fs = frozenset(L)
L1 = list(fs)

print("Frozenset:", fs)
print("Sliced Elements:", L1[1:4])
```

```python
# 9. WAP in Python to find unique elements across multiple sets.
s1 = set(map(int, input("Enter elements of set 1: ").split()))
s2 = set(map(int, input("Enter elements of set 2: ").split()))
s3 = set(map(int, input("Enter elements of set 3: ").split()))

result = s1.union(s2, s3)
print("Unique Elements:", result)
```
