```python
# 1. Write a python program to Create a string made of the middle three characters.
# str1 = JhonDipPeta
# Output : Dip

str1 = " JhonDipPeta "
mid = len(str1) // 2
result = str1[mid-1 : mid+2]
print("Output:", result)
```

```python
# 2. Write a python program to Create a new string as abbreviation of name taken as input
# Example 1: Input : Netaji Subhas Chandra Bose
# Output : N.S.C.Bose
# Example 2: Input : Niren Dutta
# Output: N.Dutta

name = input("Enter full name: ")
words = name.split()

for i in range(len(words)-1):
    print(words[i][0], end=".")
print(words[-1])
```

```python
# 3. Write a python program to Arrange string characters such that lowercase letters should come first
# Input:  PyNaTive
# Output : yaivePNT

s = input("Enter a string: ")
lower = ""
upper = ""

for ch in s:
    if ch.islower():
        lower = lower + ch
    else:
        upper = upper + ch

print("Output:", lower + upper)
```

```python
# 4. Write a python program to Count all letters, digits, and special symbols from a given string
s = input("Enter a string: ")
letters = 0
digits = 0
special = 0

for ch in s:
    if ch.isalpha():
        letters += 1
    elif ch.isdigit():
        digits += 1
    else:
        special += 1

print("Letters =", letters)
print("Digits =", digits)
print("Special Symbols =", special)
```

```python
# 5. Write a program to create a new string s3 made of the first char of s1, then the last char of s2, Next, the second char of s1 and second last char of s2, and so on.
# Any leftover chars go at the end of the result

s1 = input("Enter first string: ")
s2 = input("Enter second string: ")
s3 = ""
n = min(len(s1), len(s2))

for i in range(n):
    s3 = s3 + s1[i] + s2[-(i+1)]
s3 = s3 + s1[n:] + s2[:len(s2)-n]
print(s3)
```

```python
# 6. Write a python program to Find all occurrences of a substring in a given string by ignoring the case
# str1 = Welcome to INDIA. INDIA is  awesome
# Output: The INDIA count is: 2

str1 = input("Enter a string: ")
sub = input("Enter substring: ")
count = str1.lower().count(sub.lower())
print("Count =", count)
```

```python
# 7. Write a python program to Write a program to count occurrences of all characters within a string
s = input("Enter a string: ")
d = {}

for ch in s:
    if ch in d:
        d[ch] = d[ch] + 1
    else:
        d[ch] = 1

print("Character Count:")
print(d)
```

```python
# 8. Write a python program to Reverse a given string
s = input("Enter a string: ")
rev = ""

for ch in s:
    rev = ch + rev
print("Reversed string:", rev)
```

```python
# 9. Write a python program to Split a string on hyphens
str1 = input("Enter a string: ")
result = str1.split("-")

for word in result:
    print(word)
```

```python
# 10. Python counter and dictionary intersection example (Make a string using deletion and rearrangement). Given two strings, find if we can make first string from
# second by deleting some characters from second and rearranging remaining characters.

from collections import Counter
s1 = input("Enter first string: ")
s2 = input("Enter second string: ")

c1 = Counter(s1)
c2 = Counter(s2)

flag = True
for ch in c1:
    if c1[ch] > c2[ch]:
        flag = False
        break
if flag:
    print("Possible")
else:
    print("Not possible")
```

```python
# 11.  Write a python program to print a entered number in word pronounced

n = int(input("Enter a number: "))
d = {0:"Zero", 1:"One", 2:"Two", 3:"Three", 4:"Four", 5:"Five", 6:"Six", 7:"Seven", 8:"Eight", 9:"Nine", 10:"Ten",
     11:"Eleven", 12:"Twelve", 13:"Thirteen", 14:"Fourteen", 15:"Fifteen", 16:"Sixteen", 17:"Seventeen", 18:"Eighteen", 19:"Nineteen", 20:"Twenty",
     30:"Thirty", 40:"Forty", 50:"Fifty", 60:"Sixty", 70:"Seventy", 80:"Eighty", 90:"Ninety"}

if n >= 1000:
    print(d[n//1000], "thousand", end=" ")
    n %= 1000

if n >= 100:
    print(d[n//100], "hundred", end=" ")
    n %= 100

if n >= 20:
    print(d[(n//10)*10], d[n%10])
else:
    print(d[n])
```
