```python
#  1. Write a Python program to create a function with variable length of arguments.
def sumoflistvalues(*args):
    sum = 0
    for n in args:
        sum += n
    return sum

print("Sum of the list elements : ",sumoflistvalues(1, 2, 3, 4, 25))
```

```python
# 4. Write a Python program to create a recursive function.
def rfibo(n):
   if n <= 1:
       return n
   else:
       return(rfibo(n-1) + rfibo(n-2))
terms = int(input("Enter Term : "))
if terms <= 0:
   print("Plese enter a positive integer")
else:
   print("Fibonacci Series:")
   for i in range(terms):
       print(rfibo(i),end=" ")
```

```python
# 6. Write a Python program to reverse a string using recursive function.
# Sample String : 1234abcd , Expected Output : dcba4321

def reverse(s):
    if len(s) == 0:
        return s
    else:
        return reverse(s[1:]) + s[0]

s = input("Enter a string: ")
print("Reversed String:", reverse(s))
```

```python
#7. Write a Python function to calculate the factorial of a number (a non-negative integer). The function accepts the number as an argument.
def fact(num):
    if num == 0 or num == 1:
        return 1
    elif num < 0:
        print("Factorial not possible for negative integers!!")
        return
    else:
        f = 1
        for i in range(2, num + 1):
            f = f * i
        return f

n = int(input("Enter a number: "))
print("Factorial of number is:", fact(n))
```

```python
# 8. Write a Python function that accepts a string and counts the number of upper and lower case letters.
def count_case(s):
    upper = 0
    lower = 0

    for ch in s:
        if ch.isupper():
            upper += 1
        elif ch.islower():
            lower += 1

    print("No. of Upper case characters :", upper)
    print("No. of Lower case characters :", lower)

s = input("Enter a string: ")
count_case(s)
```

```python
# 9. Write a Python function to create and print a list where the values are the squares of numbers between 1 and 30 (both included).
def square_list():
    L = []

    for i in range(1, 31):
        L.append(i * i)
    print(L)
square_list()
```
