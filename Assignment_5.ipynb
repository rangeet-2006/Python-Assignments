```python
#1. Write a Python program to sum all the items in a list.
num = int(input("Enter the size of the list: "))
k= []
print("Enter the items you want to insert in the list:")
sum= 0

for i in range(num):
  item= int(input())
  k.append(item)
  sum= sum+item
print("Sum of all the items in a list: ",sum)
```

```python
#2. Write a Python program to get the largest number from a list.
num = int(input("Enter the size of the list: "))
k= []
print("Enter the items you want to insert in the list: ")

for i in range(num):
  item= int(input())
  k.append(item)
  largest= k[0]
  if (item > largest):
    largest= item
print("Largest number in the list: ",largest)
```

```python
#3. Write a Python program to print the numbers of a specified list after removing even numbers from it.
num = int(input("Enter the size of the list: "))
k= []
print("Enter the items you want to insert in the list: ")

for i in range(num):
  item= int(input())
  k.append(item)
  if (item % 2==0):
    k.remove(item)
print("After removing even numbers")
print("New list:",k)
```

```python
#4. Write a Python program to generate all permutations of a list in Python.
from itertools import permutations
n=int(input("Enter the limit : "))
num=[]
for i in range(n):
  a=int(input("Enter the number : "))
  num.append(a)
print(num)
print("Permutations : ")
print(list(permutations(num)))
```

```python
#5. Write a Python program to find the index of an item in a specified list.
num = int(input("Enter the size of the list: "))
k= []
print("Enter the items you want to insert in the list: ")

for i in range(num):
  item= int(input())
  k.append(item)

print("Enter the item for which you want to find the index: ",end=" ")
item= int(input())
for j in range(num):
  if(k[j]==item):
    print("Index of ",item," is: ",j)
```
