```python
# 1. Write a Python program to compute element-wise sum of given tuples
t1 = tuple(map(int, input("Enter elements in tuple1: ").split()))
print("Tuple 1: ",t1)

t2 = tuple(map(int, input("Enter elements in tuple2: ").split()))
print("Tuple 2: ",t2)

res = []
for i in range (len(t1)):
    res.append(t1[i] + t2[i])
sum = tuple(res)
print("Resultant tuple after addition: ",sum)
```

```python
# 2. Write a python program that has a nested list to store toppers details. Edit the details and reprint the details.
toppers = []
n = int(input("Enter number of toppers: "))
for i in range(n):
    name = input("Enter name: ")
    marks = int(input("Enter marks: "))
    toppers.append([name, marks])
print("Original Details:", toppers)

row = int(input("Enter topper number to edit: "))
toppers[row][0] = input("Enter new name: ")
toppers[row][1] = int(input("Enter new marks: "))
print("Updated Details:", toppers)
```

```python
# 3. Write a Python program to find the repeated items of a tuple
t1 = tuple(map(int,input("Enter the elements of a tuple: ").split()))
items = []

for i in t1:
  if t1.count(i)>1 and i not in items:
    items.append(i)

print("Repeated item in the tuple is: ",items)
```

```python
# 4. Write a Python program to Sort by Frequency of second element in Tuple List
from collections import Counter
ol = [ ("A", 5), ("B", 2), ("C", 4), ("D",3), ("C",4)]
print("Original List of tuples : " , ol)
fq = Counter(ol for key, ol in ol)
slt = sorted(ol, key = lambda fn: fq[fn[1]],reverse=True)
print("Sorted List of tuples : " , slt)
```

```python
# 5. Write a Python program to Sort lists in tuple
n = int(input("Enter number of lists: "))
L = []

for i in range (n):
  lst = list(map(int,input("Enter list elements: ").split()))
  L.append(lst)
t1 = tuple(L)
print("Original tuple: ",t1)

M = []
for lst in t1:
  lst.sort()
  M.append(lst)
t2 = tuple(M)
print("Sorted tuple: ",t2)
```

```python
# 6. Write a Python program to Order Tuples using external List
n = int (input("Enter number of tuples: "))
L = []
for i in range (n):
  ch = input("Enter character: ")
  num = int(input("Enter number: "))
  L.append((ch,num))

lst = input("Enter elements in external list: ").split()
print("Original list: ",L)
print("External list: ",lst)
L1 = []
for i in lst:
  for j in L:
    if (i == j[0]):
      L1.append(j)

print("Ordered list: ",L1)
```

```python
# 7. Write a Python program to Filter Tuples by Kth element from List
n = int (input("Enter number of tuples: "))
L = []
for i in range (n):
  tup = tuple(map(int,input("Enter tuple elements: ").split()))
  L.append(tup)
print("Original list: ",L)

k = int(input("Enter the position you want to filter: "))
val = int(input("Enter the value in that position you want to filter: "))
L1 = []
for item in L:
  if (item[k-1] == val):
    L1.append(item)
print("Filtered list: ",L1)
```

```python
# 8. Write a Python program to find a touple from a nested touple with respect to key element search which is the first element of all the nested touples .
nt = ((1, "Argentina"),(2, "Brazil"),(3, "Germany"),(4, "Spain"),(5,"Italy"),(5, "Spain"))
kv = int(input("Enter the key value to search: "))

flag = None
for item in nt:
    if item[0] == kv:
        flag = item
        break
if flag:
    print("First element of all is :", flag)
else:
    print("Tuple not found")
```

```python
# 9. Write a Python program to Sort Tuples by their Maximum element
n = int(input("Enter number of tuples: "))
L = []
for i in range(n):
    tup = tuple(map(int, input("Enter tuple elements: ").split()))
    L.append(tup)
print("Original list:", L)

for i in range(len(L)):
    for j in range(i + 1, len(L)):
        if max(L[i]) > max(L[j]):
            L[i], L[j] = L[j], L[i]
print("Sorted list:", L)
```

```python
# 10. Write a Python program to Remove nested records from tuple
it = ('Argentina', ('Brazil', 'Germany'), 'Spain', 'France', 'Uruguay')
print("Original Tuple : " , it)

dt = []
for i in it:
	if not type(i) is tuple:
		dt.append(i)

ot = tuple(dt)
print("Result after removal of nested tuple :", ot)
```
