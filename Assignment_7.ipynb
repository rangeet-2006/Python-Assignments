```python
# 1. Write a Python script to print a dictionary where the keys are numbers between 1 and n (both included) and the values are square of keys.
n = int(input("Enter the length of dictionary: "))
d = {}
for i in range(1,n+1):
  d[i] = i**2
print(d)
```

```python
# 2. Write a Python program to sum all the items in a dictionary.
n = int(input("Enter the length of dictionary: "))
d = {}
for i in range(n):
  key = input("Enter key: ")
  value = int(input("Enter value: "))
  d[key] = value
print(d)

sum = 0
for j in d.values():
    sum = sum + j
print("Sum of all the items is: ",sum)
```

```python
# 3. Write a Python program to count number of items in a dictionary value that is a list.
d = {'A': [1, 2, 3], 'B': 10, 'C': [4, 5], 'D': "Hello"}
count = 0
for i in d.values():
  if type(i) is list:
    count = count + 1
print("Number of list values = ",count)
```

```python
# 4. Write a Python program to remove duplicate values from dictionary.
n = int(input("Enter the length of dictionary: "))
d = {}
for i in range(n):
  key = int(input("Enter key: "))
  value = input("Enter value: ")
  d[key] = value
print("Original dictionary: ",d)

d1 = {}
seen = []
for k, v in d.items():
    if v not in seen:
        seen.append(v)
        d1[k] = v
print("Dictionary after removing duplicate values:", d1)
```

```python
# 5. Write a Python program to Convert two lists into a dictionary
n = int(input("Enter number of elements: "))
list1 = []
list2 = []

for i in range(n):
    list1.append(input("Enter key: "))
for i in range(n):
    list2.append(input("Enter value: "))

d = {}
for i in range(n):
    d[list1[i]] = list2[i]
print("Dictionary:", d)
```

```python
# 6. Write a Python program to Replace words from Dictionary
n = int(input("Enter the length of dictionary: "))
d = {}
for i in range(n):
  key = input("Enter key: ")
  value = input("Enter value: ")
  d[key] = value

a = input("Enter a sentence : ")
for k,v in d.items():
  a = a.replace(k,v)
print("Mosified Sentence: ",a)
```

```python
# 7. Write a program to get the maximum and minimum value of dictionary.
n = int(input("Enter the length of dictionary: "))
d = {}
for i in range(n):
  key = input("Enter key: ")
  value = int(input("Enter value: "))
  d[key] = value
print(d)

print("Maximum value: ", (max(d.values())))
print("Minimum value: ", (min(d.values())))
```

```python
# 8. Given a list of elements, perform grouping of similar elements, as different key-value lists in a dictionary.
# Input : test_list = [4, 6, 6, 4, 2, 2, 4, 8, 5, 8]
# Output : {4: [4, 4, 4], 6: [6, 6], 2: [2, 2], 8: [8, 8], 5: [5]}

test_list = [4, 6, 6, 4, 2, 2, 4, 8, 5, 8]
d = {}

for i in test_list:
    if i not in d:
        d[i] = [i]
    else:
        d[i].append(i)

print("Output : ", d)
```
